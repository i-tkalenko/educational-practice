# Часть 1. Основы

## Задача 1. Первый try/except

*Новое: `try`, `except ValueError`*

Код, который нужно проверить, помещают в блок `try`. Если там случилась ошибка указанного
типа, Python не останавливает программу, а переходит в блок `except`.

`int('три')` вызывает `ValueError`: эту строку нельзя превратить в число.

````python
for answer in ['3', 'три']:
    try:
        number = int(answer)
        print('число:', number)
    except ValueError:
        print('не число:', answer)
````

**Ваша очередь:** пользователь вводит номер новости строкой. Для каждой строки из списка
`['3', 'три', '7', '1.5']` попробуйте получить число через `int()` и вывести
заголовок `TITLES` с этим номером. Если строка не число, выведите `не число:` и саму строку.

**Должно получиться:**

```
3 Оборот компании вырос
не число: три
7 Оператор запустил сеть
не число: 1.5
```

````python
# Задача 1
for answer in ['3', 'три', '7', '1.5']:
  try:
    number = int(answer)
    print(number, TITLES[number])
  except ValueError:
    print('не число:', answer)

````

## Задача 2. Несколько except

*Новое: несколько блоков `except`, `IndexError`*

Разные ошибки перехватывают разными блоками `except`. Python проверяет их сверху вниз
и выполняет первый подходящий. Здесь возможны две ошибки: строка не число
(`ValueError`) и номер за пределами списка (`IndexError`).

````python
items = ['хлеб', 'молоко', 'сыр']

for answer in ['1', 'два', '10']:
    try:
        print(answer, '->', items[int(answer)])
    except ValueError:
        print(answer, '-> не число')
    except IndexError:
        print(answer, '-> нет такого номера')
````

**Ваша очередь:** для запросов `['2', 'пять', '15', '9']` выведите заголовок новости с этим номером.
Если запрос не число, выведите `не число`, а если новости с таким номером нет, выведите
`нет новости с номером` и запрос.

**Должно получиться:**

```
2 -> Ставка снижена
пять -> не число
15 -> нет новости с номером 15
9 -> Регионы получат субсидии
```

````python
# Задача 2
for answer in ['2', 'пять', '15', '9']:
    try:
        print(answer, '->', TITLES[int(answer)])
    except ValueError:
        print(answer, '-> не число')
    except IndexError:
        print(answer, '-> нет новостей с номером', answer)
````

## Задача 3. as error, else, finally

*Новое: `as error`, `else`, `finally`, несколько типов в скобках*

Полная схема:

- `except ... as error` сохраняет ошибку в переменную: `print(error)` покажет её текст,
  `type(error).__name__` покажет тип;
- несколько типов в одном блоке пишут в скобках: `except (ValueError, IndexError)`;
- `else` выполняется, если ошибки не было;
- `finally` выполняется всегда.

````python
for answer in ['4', 'четыре']:
    try:
        number = int(answer)
    except ValueError as error:
        print('ошибка:', type(error).__name__, '-', error)
    else:
        print('удвоенное:', number * 2)
    finally:
        print('проверено:', answer)
````

**Ваша очередь:** для запросов `['1', 'один', '20']`:

- в `try` получите заголовок `TITLES[int(запрос)]`;
- при `ValueError` или `IndexError` (один блок) выведите тип и текст ошибки;
- в `else` выведите заголовок и число слов в нём;
- в `finally` выведите `проверен запрос:` и сам запрос.

**Должно получиться:**

```
Тренер покинул клуб - 3 слова
проверен запрос: 1
ошибка: ValueError - invalid literal for int() with base 10: 'один'
проверен запрос: один
ошибка: IndexError - list index out of range
проверен запрос: 20
```

````python
# Задача 3
for answer in ['1', 'один', '20']:
    try:
        number = int(answer)
        text = TITLES[number]
    except (ValueError, IndexError) as error:
        print('ошибка:', type(error).__name__, '-', error)
    else:
        print(f'{text} - {len(text.split())} слова')
    finally:
        print('проверен запрос:', answer)
````

## Задача 4. KeyError и get

*Новое: `KeyError`, `dict.get`*

Обращение к несуществующему ключу словаря вызывает `KeyError`. Её можно перехватить, а можно
не допустить: `dict.get(ключ, 0)` вернёт `0`, если ключа нет. Простую проверку лучше
писать без `try`.

````python
prices = {'хлеб': 50, 'молоко': 90}

for item in ['хлеб', 'сыр']:
    try:
        print(item, prices[item])
    except KeyError:
        print(item, '- нет в прайсе')

print('сыр через get:', prices.get('сыр', 0))
````

**Ваша очередь:** посчитайте словарь `counts`: сколько новостей в каждой рубрике `TOPICS`. Затем для рубрик
`['спорт', 'культура', 'наука']` выведите число новостей через `counts[рубрика]`
с перехватом `KeyError` (`рубрики ... нет`). В конце выведите то же самое
одной строкой через `get`: `[counts.get(t, 0) for t in ...]`.

**Должно получиться:**

```
спорт 2
рубрики культура нет
наука 2
через get: [2, 0, 2]
```

````python
# Задача 4
counts = {}
for topic in TOPICS:
  counts[topic] = counts.get(topic, 0) + 1

search = ['спорт', 'культура', 'наука']

def check(words):
  for word in words:
    try:
        print(word, counts[word])
    except KeyError:
        print(f'рубрики {word} нет')

check(search)
print(f'через get: {[counts.get(t, 0) for t in search]}')

````

## Задача 5. Грязные данные в цикле

*Новое: `try` внутри цикла, пропуск и счётчик*

В реальных данных всегда есть «мусор»: пустые строки, опечатки, `None`. Одна плохая запись
не должна ломать обработку остальных. Для этого `try` ставят внутри цикла, тогда
плохое значение пропускаем, хорошие обрабатываем.

`int('')` вызывает `ValueError`, а `int(None)` вызывает `TypeError`, поэтому перехватываем оба.

````python
raw_scores = ['5', '3', '', 'пять', '4', None]

scores = []
skipped = 0
for value in raw_scores:
    try:
        scores.append(int(value))
    except (ValueError, TypeError):
        skipped += 1

print('оценки:', scores)
print('пропущено:', skipped)
print('средняя:', sum(scores) / len(scores))
````

**Ваша очередь:** счётчик просмотров новостей выгрузился с ошибками. Скопируйте список в ячейку
(он понадобится и в задаче 8):

```python
VIEWS = ['1200', '850', '', 'тысяча', '430', None, '2100', '15к', '990', '310']
```

`VIEWS[i]` - просмотры новости `TITLES[i]`. Посчитайте сумму просмотров по корректным
значениям. Для каждого плохого значения выведите `пропуск:`, заголовок новости и само
значение (`repr(value)`). В конце выведите, сколько новостей учтено, сколько пропущено,

**Должно получиться:**

```

пропуск: Оборот компании вырос 'тысяча'
пропуск: Телескоп передал снимки None
пропуск: Оператор запустил сеть '15к'
учтено: 6 | пропущено: 4
просмотров всего: 5880
```

````python
# Задача 5
VIEWS = ['1200', '850', '', 'тысяча', '430', None, '2100', '15к', '990', '310']

verified = 0
skipped = 0
total_views = []

for value, title in zip(VIEWS, TITLES):
  try:
    total_views.append(int(value))
    verified += 1
  except (ValueError, TypeError):
    print(f'пропуск: {title} {repr(value)}')
    total_views.append(int(0))
    skipped += 1

print(f'учтено: {verified} | пропущено: {skipped}',
      f'просмотров всего: {sum(total_views)}', sep='\n'
)
````

## Задача 6. Своя проверка raise

*Новое: `raise ValueError(...)`*

Функция может сама сообщить об ошибке командой `raise`: «с такими данными я работать
не буду». Для этого подходит встроенный `ValueError` с понятным текстом.
Ошибка «всплывает» из функции в то место, где её вызвали, и перехватывают её там.

````python
def share(part, total):
    """Доля part от total. Если total равно 0 - ошибка."""
    if total == 0:
        raise ValueError('total не может быть равен 0')
    return part / total

for part, total in [(1, 4), (3, 0)]:
    try:
        print(share(part, total))
    except ValueError as error:
        print('ошибка:', error)
````

**Ваша очередь:** напишите функцию `word_count(text)`, в которой если текст пустой или состоит из одних пробелов
(`text.strip() == ''`), она вызывает `ValueError('пустой текст')`, иначе возвращает число
слов. Вызовите её в цикле для `[TEXTS[0], '', '   ', TITLES[2]]` и перехватите ошибку -
выведите либо число слов, либо `ошибка:` и текст ошибки.

**Должно получиться:**

```
слов: 32
ошибка: пустой текст
ошибка: пустой текст
слов: 2
```

````python
# Задача 6
def word_count(text):
  if text.strip() == '':
    raise ValueError('пустой текст')
  return len(text.split())

for text in [TEXTS[0], '', ' ', TITLES[2]]:
  try:
    print('слов:', word_count(text))
  except ValueError as error:
        print('ошибка:', error)
````

## Задача 7. Файла нет

*Новое: `FileNotFoundError` в функции*

Попытка открыть файл, которого нет, вызывает `FileNotFoundError` (тема 4). Удобно спрятать чтение
в функцию, и если файла нет, она сообщает об этом и возвращает «пустой» результат,
а программа работает дальше.

````python
def read_text(name):
    """Текст файла или пустая строка, если файла нет."""
    try:
        with open(name, 'r', encoding='utf-8') as file:
            return file.read()
    except FileNotFoundError:
        print('нет файла:', name)
        return ''

with open('hello.txt', 'w', encoding='utf-8') as file:
    file.write('Привет!\n')

print(repr(read_text('hello.txt')))
print(repr(read_text('helo.txt')))
````

**Ваша очередь:** запишите заголовки `TITLES` в файл `titles.txt`, по одному на строку. Напишите функцию
`count_lines(name)`: возвращает число строк в файле, а если файла нет, выводит
`нет файла:` с именем и возвращает `0`. Вызовите её для `titles.txt` и для `titels.txt`
(опечатка в имени).

```

```

````python
# Задача 7
with open('titles.txt', 'w', encoding='utf-8') as file:
  for line in TITLES:
    file.write(line + '\n')

def count_lines(name):
    try:
        with open(name, 'r', encoding='utf-8') as file:
          count = len(file.readlines())
          return count

    except FileNotFoundError:
        print('нет файла:', name)
        return 0

for name in ['titles.txt', 'titels.txt']:
    print(f'{name}: {count_lines(name)}')
````

## Задача 8. Испорченный CSV и журнал ошибок

*Новое: файл, `try` в цикле и журнал `errors.txt` вместе*

Здесь всё вместе, как при обработке настоящих данных. Создаём CSV-файл, в котором часть
строк испорчена и читаем его. Хорошие строки считаем, а плохие записываем в журнал
`errors.txt` с номером строки и текстом ошибки, чтобы они не пропали.

Обе ошибки здесь относятся к `ValueError`: и `int('три')`, и строка без запятой
(`a, b = 'orange'.split(',')` даёт «not enough values to unpack»).

````python
with open('sales_dirty.csv', 'w', encoding='utf-8') as file:
    file.write('product,amount\n')
    for line in ['apple,5', 'banana,три', 'orange', 'apple,7', 'pear,']:
        file.write(line + '\n')

with open('sales_dirty.csv', 'r', encoding='utf-8') as file:
    lines = file.readlines()

totals = {}
done = 0
with open('errors.txt', 'w', encoding='utf-8') as log:
    for number, line in enumerate(lines[1:], start=2):     # строка 1 - заголовок
        try:
            product, amount = line.strip().split(',')
            amount = int(amount)
        except ValueError as error:
            log.write(f'строка {number}: {line.strip()} - {error}\n')
        else:
            totals[product] = totals.get(product, 0) + amount
            done += 1

print('обработано:', done, '| пропущено:', len(lines) - 1 - done)
print('итоги:', totals)
with open('errors.txt', 'r', encoding='utf-8') as log:
    print(log.read())
````

## Шпаргалка занятия

| Что нужно | Как пишется |
| --- | --- |
| перехватить ошибку | `try:` ... `except ValueError:` ... |
| несколько ошибок | несколько `except` подряд, проверяются сверху вниз |
| несколько типов одним блоком | `except (ValueError, IndexError):` |
| текст и тип ошибки | `except ValueError as error:` `print(type(error).__name__, error)` |
| если ошибки не было | `else:` |
| выполнить всегда | `finally:` |
| ключа может не быть | `d.get(key, 0)` |
| пропустить плохую запись | `try` внутри цикла и счётчик пропущенных |
| сообщить об ошибке | `raise ValueError('текст')` |
| файла нет | `except FileNotFoundError:` |
| журнал ошибок | `log.write(f'строка {number}: ... - {error}\n')` |
